```python
# @title load dataset
import kagglehub

# Download latest version
path = kagglehub.dataset_download("andrewmvd/board-games")

print("Path to dataset files:", path)
```

```python
# @title find the path to the dataset
import os

files = os.listdir(path)
print("Files in dataset directory:")
for file in files:
    print(f"- {file}")
```

```python
# @title this CSV dataset has a `sep=";"` seperator
import pandas as pd
import os

# Construct the full path to the CSV file
csv_path = os.path.join(path, "bgg_dataset.csv")

# Load the dataset with semicolon separator
df = pd.read_csv(csv_path, sep=";")

# Print information about columns, types, and non-null counts
print("=== Dataset Info ===")
display(df.info())

# Calculate missing values
print("\n=== Missing Values Count ===")
missing_values = df.isnull().sum()
display(missing_values)

# Display the first few rows for a preview
print("\n=== Dataset Preview ===")
display(df.head())
```

```python
def search_games(query_str, dataframe=df):
    """
    Searches for games where the Name column contains the specified query string.
    The search is case-insensitive.
    """
    # Filter the dataframe where the 'Name' contains the query string, ignoring case
    results = dataframe[dataframe['Name'].str.contains(query_str, case=False, na=False)]
    return results

# Search for games containing "space"
# space_games = search_games("space")

# print(f"Found {len(space_games)} games containing the word 'space'.")

# Display the first 10 results
# display(space_games[['ID', 'Name', 'Year Published', 'Rating Average', 'BGG Rank']].head(10))
```

```python
# @title black games
black_games = search_games("black")

print(f"Found {len(black_games)} games containing the word 'black'.")

# Display the first 10 results
display(black_games[['ID', 'Name', 'Year Published', 'Rating Average', 'BGG Rank']].head(10))
```

```python
# @title play_games
play_games = search_games("play")

print(f"Found {len(play_games)} games containing the word 'play'.")

# Display the first 10 results
display(play_games[['ID', 'Name', 'Year Published', 'Rating Average', 'BGG Rank']].head(10))
```

```python
# @title Search for games containing "black wall"
black_wall_games = search_games("black wall")

print(f"Found {len(black_wall_games)} games containing the word 'black wall'.")

# Display the first 10 results
display(black_wall_games[['ID', 'Name', 'Year Published', 'Rating Average', 'BGG Rank']].head(10))

print()
print(black_wall_games)
```
